```python
import numpy as np
import matplotlib.pyplot as plt
from iminuit import Minuit
from iminuit.cost import LeastSquares

# 1. GERÇEK VERİ OLUŞTURMA (Deneysel veri simülasyonu)
np.random.seed(42)
x_veri = np.linspace(0, 10, 15)
y_hata = 0.8  # Her ölçümün hata payı (sigma)

# Gerçek model: y = 2.0 * x + 3.0 + gürültü
y_veri = 2.0 * x_veri + 3.0 + np.random.normal(0, y_hata, size=len(x_veri))

# 2. MODEL FONKSİYONU
def model_dogru(x, a, b):
    return a * x + b

# 3. MALİYET FONKSİYONU OLUŞTURMA
# LeastSquares(x, y, yerr, model)
cost = LeastSquares(x_veri, y_veri, y_hata, model_dogru)

# 4. MINUIT NESNESİ VE BAŞLANGIÇ TAHMİNLERİ
m = Minuit(cost, a=1.0, b=1.0)  # Başlangıçta a=1, b=1 tahmin ediyoruz

# 5. OPTİMİZASYON VE HATA ANALİZİ
m.migrad()  # Minimum noktayı bulur
m.hesse()   # Parametrelerin hatalarını (standart sapmalarını) hesaplar

# 6. SONUÇLARI YAZDIRMA
print("Uydurma Başarılı mı?:", m.valid)
print("Ayrıntılı Parametre Tablosu:")
print(m.params)

# Veri noktalarını çizelim
plt.errorbar(x_veri, y_veri, yerr=y_hata, fmt="o", label="Deneysel Veri", capsize=3)

# Uydurulan çizgiyi çizelim
x_cizgi = np.linspace(0, 10, 100)
y_cizgi = model_dogru(x_cizgi, *m.values)  # *m.values -> a ve b parametrelerini otomatik aktarır

plt.plot(x_cizgi, y_cizgi, label=f"Fit: y = {m.values['a']:.2f}x + {m.values['b']:.2f}", color="red")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from iminuit import Minuit
from iminuit.cost import LeastSquares

np.random.seed(42)
x_veri = np.linspace(-5,5,20)
y_hata= 1.5
y_veri= 1.5*(x_veri)**2 - 0.8*(x_veri) + 2.0

def model(x,a,b,c):
    return a*x**2 +b*x + c

cost = LeastSquares(x_veri,y_veri,y_hata,model)
m = Minuit(cost,a=0,b=0,c=0)
m.fixed['c']= True
m.values['c']=2
m.limits['b']=(-0.5,0.5)
m.migrad()
m.hesse()

print(m.params)

plt.errorbar(x_veri, y_veri, yerr= y_hata , fmt='o', label='Deneysel veri')
x_line = np.linspace(-5,5,20)
y_line= model(x_veri, *m.values)
plt.plot(x_line,y_line, label=f"Fit: y = {m.values['a']:.2f}x**2 + {m.values['b']:.2f}x + {m.values['c']:.2f}", color="red")
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from iminuit import Minuit
from iminuit.cost import LeastSquares

np.random.seed(101)
x_veri= np.linspace(-5,5,40)
y_hata= 0.1
y_veri = 5.0 * np.exp(-0.5 * ((x_veri - 0.5) / 1.2)**2) + np.random.normal(0,y_hata, size= len(x_veri))

def gaussian(x,A, mu,sigma):
    return A*np.exp(-(x-mu)**2/2*sigma**2)

cost = LeastSquares(x_veri,y_veri,y_hata,gaussian)
m = Minuit(cost,A=0, mu=0, sigma=1)
m.limits["A"] = (0, None)
m.limits["sigma"] = (0.1, None)
m.migrad()
m.hesse()
print(m.params)

plt.errorbar(x_veri, y_veri, yerr=y_hata, fmt="o", label="Veri")
plt.plot(x_veri, gaussian(x_veri,*m.values), color="red", label="Gauss Fit")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from iminuit import Minuit
from iminuit.cost import LeastSquares

# 1. VERİ OLUŞTURMA (y = a * x + b)
np.random.seed(42)
x_veri = np.linspace(0, 10, 20)
y_hata = 0.5
y_veri = 2.5 * x_veri + 1.0 + np.random.normal(0, y_hata, size=len(x_veri))

# 2. MODEL VE MALİYET FONKSİYONU
def model_dogru(x, a, b):
    return a * x + b

cost = LeastSquares(x_veri, y_veri, y_hata, model_dogru)
m = Minuit(cost, a=1.0, b=0.0)

# 3. OPTİMİZASYON VE HATA HESABI
m.migrad()
m.hesse()

# 4. MATRİSLERİ ALMA
cov_matrix = m.covariance                    # Kovaryans Matrisi
corr_matrix = m.covariance.correlation()     # Korelasyon Matrisi
param_names = list(m.parameters)            # Parametre isimleri ['a', 'b']

# 5. ISI HARİTASI (HEATMAP) ÇİZİMİ
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# --- A) Kovaryans Matrisi ---
im1 = axes[0].imshow(cov_matrix, cmap="Blues")
axes[0].set_title("Kovaryans Matrisi (Covariance Matrix)")
axes[0].set_xticks(range(len(param_names)), param_names)
axes[0].set_yticks(range(len(param_names)), param_names)
fig.colorbar(im1, ax=axes[0])

# Hücrelerin üzerine sayısal değerleri yazdıralım
for i in range(len(param_names)):
    for j in range(len(param_names)):
        axes[0].text(j, i, f"{cov_matrix[i, j]:.4f}", ha="center", va="center", color="black")

# --- B) Korelasyon Matrisi ---
im2 = axes[1].imshow(corr_matrix, cmap="coolwarm", vmin=-1, vmax=1)
axes[1].set_title("Korelasyon Matrisi (Correlation Matrix)")
axes[1].set_xticks(range(len(param_names)), param_names)
axes[1].set_yticks(range(len(param_names)), param_names)
fig.colorbar(im2, ax=axes[1])

# Hücrelerin üzerine sayısal değerleri yazdıralım
for i in range(len(param_names)):
    for j in range(len(param_names)):
        axes[1].text(j, i, f"{corr_matrix[i, j]:.3f}", ha="center", va="center", color="black")

plt.tight_layout()
plt.show()
```
