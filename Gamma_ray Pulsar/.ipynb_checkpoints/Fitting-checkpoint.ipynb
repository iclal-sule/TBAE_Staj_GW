```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
# Datafile9 verilerini kullanarak fit yapma
# Read the data from the file
df = pd.read_csv('pulsar_verisi_data9file.csv')

# Convert to log values
P = np.log10(df['P'])
Pdot = np.log10(df['PdotCor'])
Ecutoff = np.log10(df['ECut_HYB'])
Edot = np.log10(df['EdotCor'])
Bdot = np.log10(df['BdotS,Cor'])
L = np.log10(df['LUMG'])

# First fitting: L ~ Ecutoff + constant
First = sm.add_constant(Ecutoff)
model1 = sm.OLS(L, First).fit()
print(model1.summary())

# Second fitting: L ~ Ecutoff + Edot + constant
X2 = pd.concat([Ecutoff, Edot], axis=1)
Second = sm.add_constant(X2)
model2 = sm.OLS(L,Second).fit()
print(model2.summary())

# Third fitting: L ~ Ecutoff + Edot + Bdot + constant
X3 = pd.concat([Ecutoff, Edot,Bdot], axis=1)
Third = sm.add_constant(X3)
model3 = sm.OLS(L,Third).fit()
print(model3.summary())
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
# Katalog verilerini okuyarak fit yapma
# Read the data from the file
df_catalog = pd.read_csv('pulsar_verisi_catalog.csv')

# Log error veriyordu sıfırdan büyük olma şartı ekledim
df_catalog = df_catalog[
    (df_catalog['LUMG'] > 0) & 
    (df_catalog['ECut_HYB'] > 0) & 
    (df_catalog['EDOTcorr'] > 0) & 
    (df_catalog['B_S'] > 0)]
    
# Convert to log values
P = np.log10(df_catalog['P0'])
Pdot = np.log10(df_catalog['P1corr'])
Ecutoff = np.log10(df_catalog['ECut_HYB'])
Edot = np.log10(df_catalog['EDOTcorr'])
Bdot = np.log10(df_catalog['B_S'])
L = np.log10(df_catalog['LUMG'])

# First fitting: L ~ Ecutoff + constant
First = sm.add_constant(Ecutoff)
model1 = sm.OLS(L, First).fit()
print(model1.summary())

# Second fitting: L ~ Ecutoff + Edot + constant
X2 = pd.concat([Ecutoff, Edot], axis=1)
Second = sm.add_constant(X2)
model2 = sm.OLS(L,Second).fit()
print(model2.summary())

# Third fitting: L ~ Ecutoff + Edot + Bdot + constant
X3 = pd.concat([Ecutoff, Edot,Bdot], axis=1)
Third = sm.add_constant(X3)
model3 = sm.OLS(L,Third).fit()
print(model3.summary())
```

Datafile9 tablosundan aldığım verilerle yapılandan çok daha güzel sonuçlar çıktı beklendiği gibi. Katalogtaki veriler düzenlenmediği için R^2 düşük geldi.
