```python
import pandas as pd
# datafile 9 dosyasındaki verileri temizleyip onlarla fit yapıcam.
# Byte-by-byte tablosuna göre sütun sınırları 
sinirlar = [
    (0, 11),   # Name (Bayt 1-11)
    (12, 17),  # f_Name (Bayt 13-17)
    (18, 26),  # LUMG (Bayt 19-26)
    (27, 33),  # P (Bayt 28-33)
    (34, 42),  # PdotCor (Bayt 35-42)
    (43, 51),  # EdotCor (Bayt 44-51)
    (52, 60),  # BdotS,Cor (Bayt 53-60)
    (61, 68),  # ECut,SED (Bayt 62-68)
    (68, 76),  # n_ECut,SED (Bayt 70-76)
    (76, 82),  # ECut,B23 (Bayt 78-82)
    (82, 90),  # n_ECut,B23 (Bayt 84-90)
    (90, 95),  # ECut,BFR (Bayt 92-95)
    (95, 103), # n_ECut,BFR (Bayt 97-103)
    (103, 109),# ECut,HYB (Bayt 105-109)
    (109, 117) # n_ECut,HYB (Bayt 111-117)
]

isimler = [
    "Name", "f_Name", "LUMG", "P", "PdotCor", "EdotCor", "BdotS,Cor", 
    "ECut,SED", "n_ECut,SED", "ECut,B23", "n_ECut,B23", 
    "ECut,BFR", "n_ECut,BFR", "ECut_HYB", "n_ECut_HYB"
]

# İlk 36 satırı atlıyoruz (skiprows=36)
df = pd.read_fwf("/home/iclal-sule-agirman/İndirilenler/datafile9.txt", colspecs=sinirlar, names=isimler, skiprows=36)

# Sütunları seçiyoruz
hedef_sutunlar = ["Name", "LUMG", "P", "PdotCor", "EdotCor", "BdotS,Cor", "ECut_HYB", "n_ECut_HYB"]
df_secili = df[hedef_sutunlar].copy()

# Sayısal dönüşüm
sayisal_sutunlar = ["LUMG", "P", "PdotCor", "EdotCor", "BdotS,Cor", "ECut_HYB", "n_ECut_HYB"]
for sutun in sayisal_sutunlar:
    df_secili[sutun] = pd.to_numeric(df_secili[sutun], errors='coerce')

# Eksik verili satırları silip temiz veriyi kaydediyoruz
df_temiz = df_secili.dropna(subset=sayisal_sutunlar)
df_temiz.to_csv("pulsar_verisi_data9file.csv", index=False)

print('done')
```

```python
import pandas as pd
# Katalog verilerini kullanıp fit yapacağım ama orada ecutoff dataları olmadığı için o verileri datafile9dan alıcam. 
# İsimlerine göre birleştirip yeni bir dosya oluşturup temizliyorum. Birleştirme yaptığım için sonda az data kalıyor.
df_excel = pd.read_excel('/home/iclal-sule-agirman/İndirilenler/Catalog.xls')
df_data9 = pd.read_csv('pulsar_verisi_data9file.csv')

df_ecut = df_data9[['Name', 'ECut_HYB']]
df_birlesik = pd.merge(df_excel, df_ecut, left_on='PSRJ', right_on='Name', how='inner')

hedef_sutunlar = ["NAME", "LUMG", "P0", "P1corr", "EDOTcorr", "B_S", "ECut_HYB", 'TSband']
df_secili = df_birlesik[hedef_sutunlar].copy()

# Sayısal dönüşüm
sayisal_sutunlar = ["LUMG", "P0", "P1corr", "EDOTcorr", "B_S", "ECut_HYB", 'TSband']
for sutun in sayisal_sutunlar:
    df_secili[sutun] = pd.to_numeric(df_secili[sutun], errors='coerce')

# Eksik verili satırları silip temiz veriyi kaydediyoruz
df_temiz = df_secili.dropna(subset=sayisal_sutunlar)
df_temiz.to_csv("pulsar_verisi_catalog.csv", index=False)


print('done')
```
